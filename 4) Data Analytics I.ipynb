```python
# 1. Import libraries
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, r2_score
import matplotlib.pyplot as plt
```

```python
# 2. Load dataset
df = pd.read_csv("boston.csv") 
```

```python
# 3. Separate Features and Target
X = df.drop("MEDV", axis=1)  # MEDV is the target column (house price)
y = df["MEDV"]
```

```python
# 4. Scale the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# 5. Split data into train and test sets
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)
```

```python
# 6. Train the model
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
# 7. Make predictions
y_pred = model.predict(X_test)
```

```python

# 8. Evaluate the model
print("Mean Squared Error:", mean_squared_error(y_test, y_pred)) #Measures average squared difference between actual and predicted values.
print("R² Score:", r2_score(y_test, y_pred)) #Closer to 1 = better. Indicates how well the features explain the target.
```

```python
# 9. Plot actual vs predicted prices
plt.scatter(y_test, y_pred, color='blue', alpha=0.6)
plt.plot([min(y_test), max(y_test)], [min(y_test), max(y_test)], color='red', linestyle='--')
plt.xlabel("Actual MEDV")
plt.ylabel("Predicted MEDV")
plt.title("Actual vs Predicted House Prices")
plt.grid(True)
plt.show()
```
