```python
# The line that you see represents the kernel density estimation. You can remove this line 
# by passing False as the parameter for the kde attribute as shown below:
sns.distplot(data['fare'], kde=False)
```

```python
# pass the value for the bins parameter in order to see more or less details in the graph
# Here we set the number of bins to 10. In the output, you will see data distributed in 10 bins
sns.distplot(data['fare'], kde=False, bins=10)
```

Distributional plot

```python
sns.jointplot(x='age', y='fare', data=titanic, kind='scatter')
plt.show()
```

```python
sns.pairplot(titanic[['age', 'fare', 'pclass', 'survived']].dropna(), hue='survived')
plt.show()
```

```python
sns.rugplot(x='fare', data=titanic)
plt.title("Rug Plot of Fare")
plt.show()
```

categorial plot

```python
sns.barplot(x='sex', y='fare', data=titanic)
plt.title("Average Fare by Gender")
plt.show()
```

```python
sns.countplot(x='class', data=titanic)
plt.title("Passenger Count by Class")
plt.show()
```

```python
sns.boxplot(x='pclass', y='fare', data=titanic)
plt.title("Fare Distribution by Passenger Class")
plt.show()
```

Histogram for Fare Distribution

```python
sns.histplot(titanic['fare'], bins=30, kde=True)
plt.title("Histogram of Fare Prices")
plt.xlabel("Fare")
plt.ylabel("Number of Passengers")
plt.show()
```
